```python
class Solution:
    def findMin(self, nums: List[int]) -> int:
        low = 0
        high=len(nums)-1

        while low <high:

            mid=(low+high)//2
            if nums[mid]> nums[high]:
                low = mid+1
            elif nums[mid]<nums[high]:
                high =mid
        return nums[low]
```

```python
def split_with_sum0_carry1(num: int):
    # For every bit set in num, both a and b must have a 1 bit.
    a = num
    b = num
    
    # XOR gives the bitwise sum without carry
    bitwise_sum = a ^ b
    # AND gives the carry bits
    carry = a & b
    
    return a, b, bitwise_sum, carry

# Example:
a, b, s, c = split_with_sum0_carry1()
print(f"Inputs: a={a}, b={b}")
print(f"Sum (a ^ b): {s}")      # Output: 0
print(f"Carry (a & b): {c}")
```

```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


def create_linked_list(arr):
    if not arr:
        return None

    head = ListNode(arr[0])
    current = head

    for value in arr[1:]:
        current.next = ListNode(value)
        current = current.next

    return head


# Input arrays
l1 = [2, 4, 3]
l2 = [5, 6, 4]

# Create linked lists
head1 = create_linked_list(l1)
head2 = create_linked_list(l2)
print(head1.val)

# Print linked list
def print_linked_list(head):
    while head:
        print(head.val, end=" -> ")
        head = head.next
    print("None")


print_linked_list(head1)
print_linked_list(head2)

def divid(total,carry):

    sum=total%10
    carry = total//10

    return sum ,carry

carry =0
sum =0



list1=[]


while head1 is not None or head2 is not None or carry>0:
    num1=head1.val if head1 else 0
    num2=head2.val if head2 else 0
    total = num1+num2+carry
    print(total)
    sum,carry = divid(total,carry)
    list1.append(sum)
    if head1:
        head1=head1.next
    if head2:
        head2=head2.next

print(list1)




```
