Maximum Subarray with one deletion 

```python
arr=[1,-2,0,3]

no_delete=arr[0]
one_delete=float("-inf")
answer=arr[0]
for i in range(1, len(arr)):

    one_delete = max(no_delete,one_delete+arr[i])

    no_delete = max(arr[i],no_delete+arr[i])

    answer=max(answer,no_delete,one_delete)

    print(answer)
```

Maximum Subarray 


```python
nums = [-2,1,-3,4,-1,2,1,-5,4]

current_sum =nums[0]
max_sum=nums[0]

for right in range(1,len(nums)):
    current_sum =max(nums[right],current_sum+nums[right])
    max_sum = max(max_sum , current_sum)

print( max_sum)
```

Maximum Absolute Subarray Sum 

```python
nums = [1,-3,2,3,-4]


min_sum =nums[0]
max_sum=nums[0]

current_min=nums[0]
current_max=nums[0]


for i in range(1,len(nums)):

    current_min = min(nums[i],current_min+nums[i])
    min_sum = min(min_sum,current_min)

    current_max= max(nums[i],current_max+nums[i])
    max_sum = max(max_sum,current_max)

print(max(abs(min_sum),abs(max_sum)))
```
