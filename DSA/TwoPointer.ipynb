3sum Closet

```python
nums=[-1,2,1,-4]
target =1
closet=[]
nums.sort()
for i in range(len(nums)):

    left =i+1
    right=len(nums)-1

    while left<right:
        total=nums[i]+nums[left]+nums[right]

        if total>target:
            right-=1
        elif total<target:
            left+=1
    closet.append(total)

for num in closet:
    if num <target:
        print(num)
    
```
