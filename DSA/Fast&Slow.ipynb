```python
    def isHappy(self, n: int) -> bool:
        
        def div(n):
            sum =0
            while (n>0):
                digit = n%10
                sum +=digit *digit
                n=n//10

            return sum 

        slow =n 
        fast =div(n)

        while fast!=1 and slow !=fast:
            slow = div(slow)
            fast = div(div(fast))
        return fast ==1
```
