```python
class Solution:
    def isPalindrome(self, head: Optional[ListNode]) -> bool:
        if not head or not head.next:
            return True
        slow,fast=head,head
        while fast and fast.next:
            slow=slow.next
            fast=fast.next.next
        re=self.reverseList(slow)
        
        while re:
            if head.val!=re.val:
                return False
            head=head.next
            re=re.next
        return True
    
    def reverseList(self, head: Optional[ListNode]) -> Optional[ListNode]:
        curr=head
        prev=None
        nex=None
        while curr:
            nex=curr.next
            curr.next=prev
            prev=curr
            curr=nex
        return prev
```

```python
class Solution:
    def removeNthFromEnd(self, head: Optional[ListNode], n: int) -> Optional[ListNode]:
        

        length =0
        curr=head
        while curr:
            length+=1
            curr=curr.next

        if length ==1 and n==1:
            return None 

        if n==length:
            return head.next
        
        slow=fast=head

        count=0
        while fast and fast.next:
            fast=fast.next
            count+=1
            if count>n:

                slow=slow.next

        if slow and slow.next:
            slow.next=slow.next.next
            

        return head
```
