Happy Number


```python
    def isHappy(self, n: int) -> bool:
        
        def div(n):
            sum =0
            while (n>0):
                digit = n%10
                sum +=digit *digit
                n=n//10

            return sum 

        slow =n 
        fast =div(n)

        while fast!=1 and slow !=fast:
            slow = div(slow)
            fast = div(div(fast))
        return fast ==1
```

```python
# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def addTwoNumbers(self, l1: Optional[ListNode], l2: Optional[ListNode]) -> Optional[ListNode]:
        

        def divid(num1,num2,carry):
            temp = num1.val+num2.val
            sum=temp%10
            carry = temp//10

            return sum, carry
        carry=0   
        sum=0
        list1=[]
        num1=l1.val
        num2=l2.val
        while l1.next !=None:
            sum,carry=divid(num1,num2,carry)
            list1.append(sum)
            num1=l1.next
            num2=l2.next
        return list1
```

```python
class Solution:
    def isPalindrome(self, head: Optional[ListNode]) -> bool:
        if not head or not head.next:
            return True
        slow,fast=head,head
        while fast and fast.next:
            slow=slow.next
            fast=fast.next.next
        re=self.reverseList(slow)
        
        while re:
            if head.val!=re.val:
                return False
            head=head.next
            re=re.next
        return True
    
    def reverseList(self, head: Optional[ListNode]) -> Optional[ListNode]:
        curr=head
        prev=None
        nex=None
        while curr:
            nex=curr.next
            curr.next=prev
            prev=curr
            curr=nex
        return prev
```
