Happy Number


```python
    def isHappy(self, n: int) -> bool:
        
        def div(n):
            sum =0
            while (n>0):
                digit = n%10
                sum +=digit *digit
                n=n//10

            return sum 

        slow =n 
        fast =div(n)

        while fast!=1 and slow !=fast:
            slow = div(slow)
            fast = div(div(fast))
        return fast ==1
```

Add Two Numbers


```python
# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def addTwoNumbers(self, l1: Optional[ListNode], l2: Optional[ListNode]) -> Optional[ListNode]:
        

        def divid(num1,num2,carry):
            temp = num1.val+num2.val
            sum=temp%10
            carry = temp//10

            return sum, carry
        carry=0   
        sum=0
        list1=[]
        num1=l1.val
        num2=l2.val
        while l1.next !=None:
            sum,carry=divid(num1,num2,carry)
            list1.append(sum)
            num1=l1.next
            num2=l2.next
        return list1
```
