```python
s = "ADOBECODEBANC"

t = "ABC"
t_count={}

for i in range(len(t)):
    t_count[t[i]]=t_count.get(t[i],0)+1

left=0
formed=0
min_len=float("inf")
window_count={}
best_left=0
best_right=0

required = len(t_count)

for right in range(len(s)):

    window_count[s[right]]=window_count.get(s[right],0)+1

    if s[right] in t_count and window_count[s[right]]==t_count[s[right]]:
        formed+=1

    while left<=right and formed==required:

        if right-left+1 < min_len:
            min_len=right-left+1
            best_left=left
            best_right=right
            print(s[best_left:best_right+1])

        left_chat=s[left]
        window_count[left_chat]-=1

        if left_chat in t_count and window_count[left_chat]<t_count[left_chat]:
            formed-=1
        left+=1
print(s[best_left:best_right+1])
```
